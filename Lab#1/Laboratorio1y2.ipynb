## 3. Preparación del corpus

Aplicamos el pipeline de normalización visto en clase, **en este orden**, y después de cada
paso contamos cuántos **tokens** y cuántos **tipos** (palabras distintas) quedan.

> **Tokens** = cada ocurrencia de una palabra, incluidas las repeticiones.
> **Tipos** = palabras distintas del vocabulario; cada tipo se cuenta una sola vez.

Guardamos **una columna nueva por paso** para poder inspeccionar y comparar cualquier etapa.

```python
def contar(serie_de_listas):
    """Devuelve (total de tokens, tamano del vocabulario) de una columna de listas."""
    total_tokens = int(serie_de_listas.str.len().sum())
    vocabulario = Counter(t for lista in serie_de_listas for t in lista)
    return total_tokens, len(vocabulario)

registro = []

def registrar(paso, serie):
    tokens, tipos = contar(serie)
    registro.append({"paso": paso, "tokens": tokens, "tipos": tipos})
    print(f"{paso:<28} tokens={tokens:>9,}   tipos={tipos:>8,}")
```

### i. Tokenización

```python
# Usamos el tokenizador POR REGLAS LINGUISTICAS de NLTK (no split por espacios),
# porque separa la puntuacion pegada a la palabra e incluye los signos ¿ ¡ del espanol.
df["tok_1_tokenizado"] = df["news"].astype(str).apply(
    lambda t: word_tokenize(t, language="spanish")
)

registrar("i. Tokenizacion", df["tok_1_tokenizado"])
df["tok_1_tokenizado"].iloc[0][:25]
```

```python
# Comparacion contra la tokenizacion ingenua por espacios en blanco
ejemplo = df["news"].iloc[0][:120]
print("TEXTO   :", ejemplo)
print()
print("split() :", ejemplo.split()[:15])
print()
print("NLTK    :", word_tokenize(ejemplo, language="spanish")[:15])
```

### ii. Minúsculas

```python
# Unificamos variantes: "Casa" y "casa" pasan a ser el mismo tipo.
# Nota: esto colapsa entidades con nombres comunes ("Banco" vs "banco").
df["tok_2_minusculas"] = df["tok_1_tokenizado"].apply(
    lambda tokens: [t.lower() for t in tokens]
)

registrar("ii. Minusculas", df["tok_2_minusculas"])
```

### iii. Eliminar puntuación

```python
# string.punctuation es ASCII: NO incluye ¿ ¡ « » — “ ” …
# Por eso ampliamos el conjunto a mano (desafio propio del espanol).
PUNTUACION = set(string.punctuation) | set("¿¡«»—–“”‘’…´`")
print("Signos considerados:", "".join(sorted(PUNTUACION)))

# Medimos cuantos tokens numericos hay, para saber que perdemos con isalpha()
todos = [t for lista in df["tok_2_minusculas"] for t in lista]
print(f"\nTokens puramente numericos: {sum(1 for t in todos if t.isdigit()):,}")
print(f"Tokens de solo puntuacion : {sum(1 for t in todos if all(c in PUNTUACION for c in t)):,}")
del todos
```

```python
# Criterio elegido: conservamos solo tokens alfabeticos (isalpha()).
# Esto elimina signos de puntuacion Y numeros; lo documentamos en el reporte.
df["tok_3_sin_punt"] = df["tok_2_minusculas"].apply(
    lambda tokens: [t for t in tokens if t.isalpha()]
)

registrar("iii. Sin puntuacion", df["tok_3_sin_punt"])
```

### iv. Eliminar stopwords

```python
# Lista de stopwords del espanol de NLTK. La convertimos a SET:
# con una lista, el operador "in" es O(n) por token y sobre cientos de miles se nota.
STOPWORDS_ES = set(stopwords.words("spanish"))
print(f"Stopwords del espanol en NLTK: {len(STOPWORDS_ES)}")
print(sorted(STOPWORDS_ES)[:30])

# Nota: vienen CON tilde ("está", "más", "sí"). Si se quitaran los acentos del corpus,
# habria que quitarselos tambien a esta lista o no filtrarian nada.
print("\n¿'está' en la lista?", "está" in STOPWORDS_ES)
print("¿'esta' en la lista? ", "esta" in STOPWORDS_ES)
```

```python
df["tok_4_sin_stop"] = df["tok_3_sin_punt"].apply(
    lambda tokens: [t for t in tokens if t not in STOPWORDS_ES]
)

registrar("iv. Sin stopwords", df["tok_4_sin_stop"])
```

```python
# Efecto de quitar stopwords sobre el SIGNIFICADO: la negacion "no" es stopword.
print("'no' es stopword: ", "no" in STOPWORDS_ES)
print("'ni' es stopword: ", "ni" in STOPWORDS_ES)
print("'sin' es stopword:", "sin" in STOPWORDS_ES)

# Buscamos una noticia real que contenga una negacion y vemos como queda
for i in range(len(df)):
    if "no" in df["tok_3_sin_punt"].iloc[i][:40]:
        print("\nANTES :", df["tok_3_sin_punt"].iloc[i][:25])
        print("DESPUES:", df["tok_4_sin_stop"].iloc[i][:25])
        break
```

### v. Lematización

**Importante:** el `WordNetLemmatizer` de NLTK es **solo para inglés**; aplicado al español
devolvería las palabras casi intactas. Usamos **spaCy** con el modelo `es_core_news_sm`,
que es el que reduce cada palabra a su forma de diccionario usando información gramatical.

```python
import spacy
from spacy.tokens import Doc

# Cargamos el modelo del espanol. Desactivamos parser y NER (no los necesitamos)
# pero SI dejamos tagger/morphologizer/attribute_ruler: el lematizador los requiere.
nlp = spacy.load("es_core_news_sm", disable=["parser", "ner"])
print("Componentes activos:", nlp.pipe_names)
```

```python
# Respetamos nuestra tokenizacion previa construyendo Doc a partir de la lista de tokens,
# en lugar de re-tokenizar con spaCy.
docs = [Doc(nlp.vocab, words=tokens) for tokens in df["tok_4_sin_stop"]]

lemas = []
for doc in nlp.pipe(docs, batch_size=64):
    lemas.append([token.lemma_.lower() for token in doc])

df["tok_5_lemas"] = lemas
registrar("v. Lematizacion", df["tok_5_lemas"])
```

### 5.2 Riqueza léxica (razón tipo/token)

```python
ttr_global = tipos_desp / tokens_desp
print(f"TTR global (normalizado): {ttr_global:.5f}")

filas = []
for categoria, grupo in df.groupby("Type"):
    tk, tp = contar(grupo["tok_5_lemas"])
    filas.append({
        "categoria": categoria,
        "documentos": len(grupo),
        "tokens": tk,
        "tipos": tp,
        "TTR": round(tp / tk, 5) if tk else 0,
    })

ttr_categoria = pd.DataFrame(filas).sort_values("TTR", ascending=False)
ttr_categoria
```

> **Advertencia metodológica:** el TTR baja mecánicamente conforme crece el número de
> tokens, así que comparar categorías de tamaños muy distintos está sesgado.
> Abajo repetimos el cálculo submuestreando todas las categorías al mismo número de tokens.

```python
# TTR comparable: truncamos cada categoria al mismo numero de tokens
minimo = ttr_categoria["tokens"].min()
print(f"Truncando todas las categorias a {minimo:,} tokens\n")

filas = []
for categoria, grupo in df.groupby("Type"):
    tokens_cat = [t for lista in grupo["tok_5_lemas"] for t in lista][:minimo]
    filas.append({
        "categoria": categoria,
        "tokens": len(tokens_cat),
        "tipos": len(set(tokens_cat)),
        "TTR_comparable": round(len(set(tokens_cat)) / len(tokens_cat), 5),
    })

pd.DataFrame(filas).sort_values("TTR_comparable", ascending=False)
```

### 5.5 Ley de Zipf

Graficamos frecuencia contra rango en escala log-log y la comparamos con la curva teórica
$f(rango) \approx k / rango$, donde $k$ es la frecuencia de la palabra de rango 1.

```python
def datos_zipf(contador):
    """Devuelve (rangos, frecuencias observadas, frecuencias teoricas k/rango)."""
    frecs = [f for _, f in contador.most_common()]
    rangos = list(range(1, len(frecs) + 1))
    k = frecs[0]                      # constante = frecuencia del rango 1
    teorica = [k / r for r in rangos]
    return rangos, frecs, teorica

# Tambien calculamos Zipf sobre el corpus SIN normalizar (solo tokenizado + minusculas)
# para ver el efecto de quitar las stopwords sobre la cabeza de la curva.
frecuencias_crudo = Counter(t for lista in df["tok_2_minusculas"] for t in lista)

r_n, f_n, t_n = datos_zipf(frecuencias)          # normalizado
r_c, f_c, t_c = datos_zipf(frecuencias_crudo)    # crudo
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

ax[0].loglog(r_c, f_c, color="steelblue", label="Observada (corpus crudo)")
ax[0].loglog(r_c, t_c, "--", color="black", label="Teorica  k / rango")
ax[0].set_title("Ley de Zipf — corpus SIN normalizar")
ax[0].set_xlabel("Rango (log)")
ax[0].set_ylabel("Frecuencia (log)")
ax[0].legend()
ax[0].grid(alpha=.3, which="both")

ax[1].loglog(r_n, f_n, color="indianred", label="Observada (normalizado)")
ax[1].loglog(r_n, t_n, "--", color="black", label="Teorica  k / rango")
ax[1].set_title("Ley de Zipf — corpus NORMALIZADO")
ax[1].set_xlabel("Rango (log)")
ax[1].legend()
ax[1].grid(alpha=.3, which="both")

plt.tight_layout()
plt.savefig(FIG / "04_zipf.png", dpi=150)
plt.show()
```

```python
# Evidencia numerica para el analisis:
# 1) verificamos que rango x frecuencia sea aproximadamente constante
print("rango x frecuencia (corpus normalizado):")
for r in [1, 5, 10, 50, 100, 500, 1000]:
    if r <= len(f_n):
        print(f"  rango {r:>5}  frec {f_n[r-1]:>7,}  producto {r*f_n[r-1]:>10,}")

# 2) hapax legomena: palabras que aparecen UNA sola vez
hapax = sum(1 for c in frecuencias.values() if c == 1)
print(f"\nHapax legomena: {hapax:,} de {len(frecuencias):,} tipos "
      f"({hapax/len(frecuencias)*100:.1f}% del vocabulario)")
```

## 1. Bolsa de palabras (Bag of Words)

`CountVectorizer` recibe **texto plano**, no listas de tokens, así que primero reconstruimos
cada documento uniendo sus lemas con espacios. Así el vectorizador solo tiene que separar por
espacios y no vuelve a aplicar reglas de tokenización que ya resolvimos en el Lab #1.


```python
# Texto normalizado por documento: los lemas del Lab #1 unidos en un solo string
df["texto_norm"] = df["tok_5_lemas"].str.join(" ")

df[["Type", "texto_norm"]].head(3)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Sin parametros: queremos el vocabulario COMPLETO del corpus ya normalizado.
# El texto entra pre-limpio del Lab #1, asi que el vectorizador solo separa por espacios.
vectorizador = CountVectorizer()

# fit_transform = aprende el vocabulario (fit) y construye la matriz de conteos (transform)
X_bow = vectorizador.fit_transform(df["texto_norm"])

n_docs, n_vocab = X_bow.shape
print(f"Matriz BoW: {n_docs:,} documentos x {n_vocab:,} palabras de vocabulario")
print(f"Celdas totales: {n_docs * n_vocab:,}  |  celdas con valor != 0: {X_bow.nnz:,}")

# Comparacion con el vocabulario contado a mano en el Lab #1
print(f"\nVocabulario en el Lab #1: {len(frecuencias):,}")
print(f"Vocabulario de CountVectorizer: {n_vocab:,}  diferencia: {len(frecuencias) - n_vocab:,}")
```

**Qué se pierde y qué se gana con BoW** 

Lo que se pierde es el orden, la sintaxis, el contexto y la relación en cada uno ya que se dan vectores idénticos haciendo que se pierda el sentido acorde a la estructura de la oración, como bag of words toma todas las dimensiones de forma independiente muchas palabras importantes pierden relación entre sí.


Lo que se gana es una representación numérica de tamaño fijo que permite compararla entre sí sin importar el tamaño de la noticia. Es más baráta de consumir en cuento a costo computacional ya que con una pasada sobre el corpus ya se puede interpretar con cada columna correspondiendo a cada palabra en concreto para ver directamente que se relaciona con qué en cada documento.


## 2. Vectores de conteo y dispersión

La matriz BoW tiene una celda por cada par (documento, palabra del vocabulario), pero un
documento solo activa las palabras que realmente contiene. La **dispersión** (*sparsity*) mide
qué porcentaje de esas celdas quedan en cero.


```python
# X_bow ya nos da los dos numeros necesarios:
#   X_bow.shape -> (documentos, vocabulario)
#   X_bow.nnz   -> "number of non-zero", celdas con conteo distinto de cero

# TO-DO 1: total de celdas de la matriz
total_celdas = X_bow.shape[0] * X_bow.shape[1]

# TO-DO 2: porcentaje de celdas que son CERO
sparsity = (total_celdas - X_bow.nnz) / total_celdas * 100

print(f"Celdas totales : {total_celdas:,}")
print(f"Celdas no cero : {X_bow.nnz:,}")
print(f"Dispersion     : {sparsity:.2f}% de las celdas son cero")

# Contexto: cuantas palabras distintas usa un documento promedio
print(f"\nPalabras distintas por documento (promedio): {X_bow.nnz / X_bow.shape[0]:.1f}")
```

Ahora repetimos el cálculo limitando el vocabulario a las *k* palabras más frecuentes,
para ver cómo se mueve la dispersión al agrandar el vocabulario.


```python
tamanos = [100, 500, 1000, 5000, None]   # None = vocabulario completo
resultados = []

for k in tamanos:
    # TO-DO 3: crea un CountVectorizer limitado a las k palabras MAS FRECUENTES
    #         (busca el parametro max_features) y ajustalo sobre df["texto_norm"]
    vec = CountVectorizer(max_features=k)
    X = vec.fit_transform(df["texto_norm"])

    # TO-DO 4: dispersion de esta matriz, igual que en la celda anterior
    disp = (X.shape[0] * X.shape[1] - X.nnz) / (X.shape[0] * X.shape[1]) * 100

    resultados.append({
        "vocabulario": X.shape[1],
        "celdas": X.shape[0] * X.shape[1],
        "no_cero": X.nnz,
        "dispersion_%": round(disp, 2),
    })

tabla_sparsity = pd.DataFrame(resultados)
tabla_sparsity
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))

ax[0].plot(tabla_sparsity["vocabulario"], tabla_sparsity["dispersion_%"],
           marker="o", color="steelblue")
ax[0].set_title("Dispersion vs tamano del vocabulario")
ax[0].set_xlabel("Palabras en el vocabulario")
ax[0].set_ylabel("% de celdas en cero")
ax[0].grid(alpha=.3)

# Escala log en X: los tamanos van de 100 a ~20,000 y en lineal se amontonan
ax[1].plot(tabla_sparsity["vocabulario"], tabla_sparsity["dispersion_%"],
           marker="o", color="indianred")
ax[1].set_xscale("log")
ax[1].set_title("Lo mismo con eje X logaritmico")
ax[1].set_xlabel("Palabras en el vocabulario (log)")
ax[1].grid(alpha=.3, which="both")

for a in ax:
    for _, r in tabla_sparsity.iterrows():
        a.annotate(f'{r["dispersion_%"]:.1f}%',
                   (r["vocabulario"], r["dispersion_%"]),
                   textcoords="offset points", xytext=(0, 8), ha="center", fontsize=8)

plt.tight_layout()
plt.savefig(FIG / "06_sparsity.png", dpi=150)
plt.show()
```

**Por qué crece la dispersión y qué implica en almacenamiento y cómputo**

La dispersión crece ya que las palabras que entran al ampliar el vocabulario son cada vez menos comúnes, esto implica mayor uso de almacenamiento y cómputo al recorrer varias veces el mismo campo no'cero que ya se sabe que es escalando el costo.


**Qué es una matriz dispersa y por qué se usa el formato CSR(Compressed Sparse Row)** 

Una matriz densa almacena todo en memoria sin importar si es cero o no, mientras que una matriz dispersa se construye principalmente de ceros guardando únicamente posiciones ahorrando tiempo de ejecución y espacio en memoria. 

El formato Compressed Sparse Row se tuiliza para optimizar el almacenamiento en memoria en la multiplicacion de matriz por vector, a diferencia de guardar la matriz celda por celda se divide la información en 3 arrays unidemensionales teniendo, valores, índices de columna y punteros de fila.  


## 3. N-gramas

Un n-grama es una secuencia de *n* palabras contiguas. `CountVectorizer` los genera con
`ngram_range=(min, max)`: `(1,1)` son unigramas (la bolsa de palabras clásica), `(2,2)` solo
bigramas, `(1,2)` unigramas y bigramas juntos. La idea es recuperar parte del contexto local que
BoW había tirado, y el precio es el tamaño del vocabulario.


```python
# La formula de dispersion ya la usamos dos veces en la Seccion 2 y aqui la
# necesitamos cuatro veces mas, asi que la extraemos a una funcion.
def dispersion(M):
    """% de celdas en cero de una matriz dispersa."""
    return (M.shape[0] * M.shape[1] - M.nnz) / (M.shape[0] * M.shape[1]) * 100


#respuesta a seccion 6 unigramas, bigramas, trigramas y uni+bi.
configs = {
    "unigramas (1,1)": (1, 1),
    "bigramas (2,2)":  (2, 2),
    "trigramas (3,3)": (3, 3),
    "uni+bi (1,2)":    (1, 2),
}

filas, matrices = {}, {}

for nombre, rango in configs.items():
    # TO-DO 1: crea el CountVectorizer con ngram_range=rango y ajustalo sobre df["texto_norm"]
    vec = CountVectorizer(ngram_range=rango)
    X = vec.fit_transform(df["texto_norm"])

    matrices[nombre] = (vec, X)
    filas[nombre] = {
        "vocabulario": X.shape[1],
        "no_cero": X.nnz,
        "dispersion_%": round(dispersion(X), 2),   # reutilizamos la funcion de la Seccion 2
    }

tabla_ngramas = pd.DataFrame(filas).T
tabla_ngramas
```

```python
import numpy as np

vec_bi, X_bi = matrices["bigramas (2,2)"]

# TO-DO 2: suma los conteos de cada bigrama sobre TODOS los documentos.
#         X_bi.sum(axis=0) da una matriz 1 x V; aplanar con np.asarray(...).ravel()
total_por_bigrama = np.asarray(X_bi.sum(axis=0)).ravel()

# TO-DO 3: recupera los nombres de las columnas del vectorizador
#         (el metodo de sklearn que devuelve las features)
nombres = vec_bi.get_feature_names_out()

top15_bigramas = (
    pd.DataFrame({"bigrama": nombres, "frecuencia": total_por_bigrama})
      .sort_values("frecuencia", ascending=False)
      .head(15)
      .reset_index(drop=True)
)
top15_bigramas
```

**¿Aportan los bigramas información que los unigramas no capturan?**

Sí aportan, aunque no todos ya que varios pares de palabras no dan suficiente información que igualmente no se captura en un unigrama. Varios bigramas del top-15 nombran cosas que un unigrama no puede representar, como banco central, punto básico o cambio climático, donde las palabras por separado son ambiguas o directamente no significan nada en ese contexto. También aparecen entidades como bbva research y bbva méxico, que solo existen como par. En cambio otros bigramas del mismo top, como poder ser, cada vez o hacer él, son combinaciones que no agregan tema y que sobrevivieron a la normalización, así que ordenar los bigramas por frecuencia bruta mezcla los que sirven con los que no. Además, como el texto ya viene sin stopwords, pares como tasa interés o precio consumidor no eran contiguos en la noticia original sino que quedaron pegados al eliminar las palabras vacías.




**Ventaja de los bigramas/trigramas frente a los unigramas, y su costo en vocabulario**

Dan orden, mayor contexto pero su costo computacional es mayor.
Los bigramas y trigramas recuperan parte del orden y del contexto local que la bolsa de palabras descarta, lo que permite distinguir un término compuesto de sus palabras sueltas. El costo está en el vocabulario, que pasa de 19,767 features con unigramas a 208,674 con bigramas y a 274,711 con trigramas, mientras que las celdas ocupadas apenas suben de 202,700 a 284,482 y 294,494. La dispersión sube en consecuencia de 99.10% a 99.88% y 99.91%. Junto con eso, la frecuencia promedio por feature cae de 15.1 en unigramas a 1.43 en bigramas y 1.08 en trigramas, es decir que casi cada trigrama aparece una sola vez y no sirve para generalizar. Por eso conviene combinar unigramas con bigramas y filtrar los más raros con min_df, mientras que los trigramas no se justifican en un corpus de 1,137 documentos.




## 4. TF-IDF

En BoW el peso de una palabra es su conteo, así que las palabras que aparecen en todas las
noticias dominan. TF-IDF corrige eso multiplicando dos factores: **TF**, cuánto aparece la palabra
*dentro* del documento, e **IDF**, qué tan rara es la palabra *en el corpus*. Una palabra que sale
en todos los documentos tiene IDF cercano a cero y queda anulada; una que se repite mucho en un
solo documento se dispara.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TO-DO 1: crea el vectorizador TF-IDF y ajustalo sobre df["texto_norm"]
#         (mismos parametros por defecto que CountVectorizer, solo cambia el peso de la celda)
vec_tfidf = TfidfVectorizer()
X_tfidf = vec_tfidf.fit_transform(df["texto_norm"])

print("Matriz TF-IDF:", X_tfidf.shape)
print(f"Dispersion   : {dispersion(X_tfidf):.2f}%")
print("\nMismo vocabulario que BoW:", X_tfidf.shape[1] == X_bow.shape[1])
print("Lo unico que cambia es QUE se guarda en cada celda, no cuales celdas se ocupan.")
```

```python
# Tomamos el primer documento de las 3 categorias mas frecuentes.
# Usamos POSICIONES (0..n-1) porque las filas de la matriz siguen el orden de df.
tipos = df["Type"].to_numpy()
tres_categorias = df["Type"].value_counts().index[:3].tolist()
posiciones = {cat: int(np.where(tipos == cat)[0][0]) for cat in tres_categorias}

print("Documentos elegidos (categoria -> fila):", posiciones)

terminos = vec_tfidf.get_feature_names_out()

def top_n(matriz, fila, n=10):
    """Devuelve las n columnas con mayor valor en una fila de una matriz dispersa."""
    valores = matriz[fila].toarray().ravel()
    idx = valores.argsort()[::-1][:n]
    return list(zip(terminos[idx], valores[idx]))
```

```python
# Comparacion: top-10 por TF-IDF vs top-10 por conteo simple, para los mismos documentos
for cat, fila in posiciones.items():
    # TO-DO 2: obten el top-10 de esa fila en X_tfidf y en X_bow usando top_n()
    por_tfidf  = top_n(X_tfidf, fila)
    por_conteo = top_n(X_bow, fila)

    comp = pd.DataFrame({
        "palabra (TF-IDF)": [p for p, _ in por_tfidf],
        "tf-idf":           [round(v, 3) for _, v in por_tfidf],
        "palabra (conteo)": [p for p, _ in por_conteo],
        "conteo":           [int(v) for _, v in por_conteo],
    })
    print(f"\n{'='*70}\n{cat}  (documento en la fila {fila})\n{'='*70}")
    display(comp)

    comunes = {p for p, _ in por_tfidf} & {p for p, _ in por_conteo}
    print(f"Palabras en ambos top-10: {len(comunes)}/10 -> {sorted(comunes)}")
```

**¿Qué papel juegan estas palabras en los artículos y cómo lo refleja su puntaje?**

En los tres documentos las palabras mejor puntuadas son las que definen el tema concreto de la
noticia, no las que les dan sentido gramaticalmente. En el artículo de Macroeconomía encabezan
*marzo* (0.289), *ipc* (0.263) y *tasa* (0.250), junto con términos técnicos como *ipca* e
*interanual*: el documento es una nota sobre la publicación mensual del índice de precios, y esas
palabras son justamente las que permiten identificarlo. En el de Alianzas dominan *vivo* (0.230),
*aerobus* (0.225), *aerolínea* (0.213) y *aviación* (0.191)
En el de Innovación el primer lugar es *agile* (0.422), seguido de *nacho* (0.330) y *podcast*
(0.267), que corresponden al formato y al protagonista de la pieza.

El puntaje refleja ese papel porque cae en las dos condiciones que hacen distintiva a una
palabra: que se repita dentro del documento y que sea poco común en el resto del corpus. *Agile*
alcanza 0.422 porque aparece tres veces en un texto corto y casi no aparece en las demás noticias,
mientras que palabras igual de frecuentes pero repartidas por todo el corpus quedan muy abajo. Un
caso es el de *talentboost*, *thinktank* y *nachir*, que aparecen una sola vez y aun
así empatan en 0.165: al ser exclusivas de ese documento reciben el IDF máximo posible.


**Comparación con el conteo simple: ¿coinciden? ¿Por qué ocurren las diferencias?**

Coinciden solo en parte. El solapamiento entre ambos top-10 es de 6 de 10 palabras en los
documentos de Macroeconomía y Alianzas, y de 7 de 10 en el de Innovación, de modo que entre tres y
cuatro palabras cambian en cada caso, y además el orden se reordena por completo.

Las diferencias se explican por el factor IDF, que es lo único que separa a las dos
representaciones. El conteo simple ordena por TF, así que premia a cualquier palabra repetida sin
preguntarse si es característica del documento. En el artículo de Alianzas la palabra más frecuente
es *año*, con cuatro apariciones, y sin embargo desaparece del top-10 de TF-IDF; lo mismo ocurre
con *bajo*, *trabajar* y *nuevo*, que son vocabulario general presente en gran parte del corpus y
por lo tanto reciben un IDF cercano a 1. En su lugar entran *aerobus*, *low*, *avión* y *pasajero*,
que aparecen menos veces dentro del documento pero casi no aparecen fuera de él. El artículo de
Macroeconomía muestra el mismo efecto de forma más sutil: *tasa* es la palabra más frecuente, con
ocho apariciones, pero cae al tercer lugar en TF-IDF porque es común en un corpus de noticias
económicas, mientras que *ipca*, *interanual* e *impuesto* suben pese a tener conteos bajos.

En términos de las fórmulas, el TF de dos palabras puede ser idéntico y aun así sus pesos finales
diferir en un orden de magnitud, porque el IDF depende de en cuántos documentos aparece cada una y
no de cuántas veces aparece en este.


**Por qué TF-IDF penaliza las palabras muy comunes y favorece las raras pero repetidas**

El componente IDF, definido en scikit-learn como ln((1 + N) / (1 + df)) + 1, depende únicamente de
en cuántos documentos aparece la palabra. Si una palabra está presente en las 1,137 noticias, el
cociente vale 1, su logaritmo es cero y el IDF se reduce a 1, el valor mínimo posible: esa palabra
solo conserva su frecuencia bruta y no recibe ningún refuerzo. Si en cambio aparece en cinco
documentos, el IDF ronda 6.25, de manera que una sola aparición pesa más que seis o siete apariciones de la
palabra ubicua.

La lógica detrás es que una palabra presente en todos los documentos no sirve para distinguir
ninguno. No aporta información porque no discrimina, por más que se repita. En cambio una palabra
que aparece en pocos documentos pero se repite dentro de uno señala de qué trata ese documento en
particular: el TF confirma que no es una mención accidental y el IDF confirma que es específica.
El producto de ambos factores exige las dos condiciones a la vez, y por eso el ranking resultante
se parece mucho más a un resumen del contenido que el ranking por frecuencia.

Conviene notar el límite del método: las palabras que aparecen en un único documento reciben el
IDF máximo y quedan todas empatadas, como ocurre con *talentboost*, *thinktank* y *nachir* en
0.165. Dado que en este corpus los hapax representan una fracción muy alta del vocabulario, en la
práctica suele filtrarse ese extremo con un `min_df` mayor que uno.


## 5. Similitud entre documentos (similitud coseno)

La similitud coseno mide el **ángulo** entre dos vectores, no la distancia entre sus puntas.
Va de 0 (sin palabras en común) a 1 (misma dirección). Como `TfidfVectorizer` ya devuelve los
vectores normalizados a norma 1, el coseno es directamente su producto punto.


```python
from sklearn.metrics.pairwise import cosine_similarity

# TO-DO 3: matriz de similitud coseno entre TODOS los pares de documentos TF-IDF
S = cosine_similarity(X_tfidf)

print("Matriz de similitud:", S.shape)
print(f"Diagonal (cada doc consigo mismo): {S[0, 0]:.4f}")
```

```python
# Documento de referencia: el primero de la categoria mas frecuente
ref = posiciones[tres_categorias[0]]
print(f"Documento de referencia: fila {ref}  |  categoria: {tipos[ref]}")
print(f"Texto (primeras 300 letras):\n{df['texto_norm'].iloc[ref][:300]}...\n")

# TO-DO 4: encuentra las 5 filas mas similares al documento de referencia.
#         Cuidado: la mas similar es el propio documento (similitud 1.0), hay que excluirlo.
#         Pista: S[ref].argsort()[::-1] ordena de mayor a menor.
orden = S[ref].argsort()[::-1]
top5 = orden[orden != ref][:5]   # quitamos el propio documento

pd.DataFrame({
    "fila": top5,
    "categoria": tipos[top5],
    "similitud": S[ref, top5].round(4),
})
```

```python
# Similitud promedio DENTRO de la misma categoria vs ENTRE categorias distintas.
# Usamos solo el triangulo superior para no contar cada par dos veces ni la diagonal.
iu = np.triu_indices_from(S, k=1)
sims = S[iu]
misma_cat = (tipos[:, None] == tipos[None, :])[iu]

print(f"Pares totales            : {len(sims):,}")
print(f"Similitud media INTRA-categoria: {sims[misma_cat].mean():.4f}  ({misma_cat.sum():,} pares)")
print(f"Similitud media INTER-categoria: {sims[~misma_cat].mean():.4f}  ({(~misma_cat).sum():,} pares)")
```

```python
# Extra para la Seccion 6: similitud media entre cada par de categorias.
# La diagonal es la cohesion interna de la categoria; fuera de la diagonal, la confusion.
cats = sorted(df["Type"].unique())
M = pd.DataFrame(index=cats, columns=cats, dtype=float)

for a in cats:
    for b in cats:
        ia, ib = np.where(tipos == a)[0], np.where(tipos == b)[0]
        bloque = S[np.ix_(ia, ib)]
        if a == b:                       # excluimos la diagonal (doc consigo mismo)
            n = len(ia)
            M.loc[a, b] = (bloque.sum() - n) / (n * n - n) if n > 1 else np.nan
        else:
            M.loc[a, b] = bloque.mean()

M.round(4).style.background_gradient(cmap="Blues")
```

**¿Coinciden las categorías de los 5 documentos más similares con la del documento de referencia?**

Coinciden las cinco. El documento de referencia es la fila 3, de categoría Macroeconomía, y los
cinco vecinos más cercanos son las filas 1131 (0.4694), 1124 (0.3132), 796 (0.2768), 121 (0.2580)
y 88 (0.2555), todos de Macroeconomía. El resultado es coherente con el contenido: la noticia trata
sobre la publicación del IPC, y los documentos recuperados comparten ese vocabulario técnico.

La similitud coseno separa las
categorías en promedio, pero el margen entre ambas distribuciones es bajo.


**Por qué se usa similitud coseno y no distancia euclidiana para documentos de distinta longitud**

La distancia euclidiana mide documentos de distinta longitud porque mide la distancia en línea recta entre dos puntos en función de la frecuencia absoluta de las palabras.

La similitud coseno mide el ángulo entre los vectores y no su longitud, dos documentos
con la misma proporción de palabras obtienen similitud 1 aunque uno sea mucho más largo que el
otro.

**¿Para qué sirve cada conjunto y por qué no se debe usar el de prueba durante el desarrollo?**

El corpus quedó dividido en 795 documentos de entrenamiento (69.9%), 171 de validación (15.0%) y
171 de prueba (15.0%), manteniendo en los tres la proporción original de las siete categorías.

El conjunto de **entrenamiento** es el único del que el modelo aprende: sobre él se estiman las
probabilidades que definen al clasificador. El de **validación** sirve para tomar decisiones
durante el desarrollo, como elegir entre la representación de bolsa de palabras y TF-IDF, o fijar
el tamaño del vocabulario; se mide el desempeño sobre él, se ajusta lo que haga falta y se vuelve a
medir tantas veces como sea necesario. El de **prueba** se reserva para una única medición al
final, cuando ya no queda ninguna decisión por tomar, y su función es estimar cómo se comportaría
el modelo frente a noticias que nunca ha visto.

La razón de no tocar el conjunto de prueba durante el desarrollo es que cada decisión tomada a
partir de un conjunto lo contamina como estimador imparcial. Si se probaran varias configuraciones
sobre prueba y se eligiera la de mejor resultado, ese resultado dejaría de medir la capacidad de
generalización del modelo y pasaría a medir qué tan bien se acertó al elegir para ese conjunto en
particular. El desempeño reportado sería optimista y no se sostendría con datos nuevos. Mantener
prueba intacto es lo que permite que la cifra final signifique algo, y por eso existe validación:
absorbe el desgaste de las decisiones repetidas para que prueba conserve su valor.


## 2. Construcción del clasificador Naive Bayes

```python
# Vectorizacion: fit SOLO con entrenamiento, transform para validacion y prueba.
# Dos representaciones: CountVectorizer (BoW) y TfidfVectorizer.
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

#fit aprende el vocabluario y transform construye la matriz de conteos (o TF-IDF)

vec_count = CountVectorizer()
Xtrain_bow = vec_count.fit_transform(X_train)
Xval_bow = vec_count.transform(X_val)
Xtest_bow = vec_count.transform(X_test)

vec_tfidf = TfidfVectorizer()
Xtrain_tfidf = vec_tfidf.fit_transform(X_train)
Xval_tfidf = vec_tfidf.transform(X_val)
Xtest_tfidf = vec_tfidf.transform(X_test)
print("BoW:   ", Xtrain_bow.shape, Xval_bow.shape, Xtest_bow.shape)
print("TF-IDF:   ", Xtrain_tfidf.shape, Xval_tfidf.shape, Xtest_tfidf.shape)
```

```python
# Entrenar un MultinomialNB con cada representacion.
nb_bow = MultinomialNB().fit(Xtrain_bow, y_train)
nb_tfidf = MultinomialNB().fit(Xtrain_tfidf, y_train)

print("Categorias:", nb_bow.classes_)
print("P(w|c) shape:", nb_bow.feature_log_prob_.shape, "= (categorias, vocabulario)")

# P(c): el prior aprendido vs la proporcion real en entrenamiento
pd.DataFrame({
    "P(c) del modelo": np.exp(nb_bow.class_log_prior_).round(4),
    "proporcion en train": y_train.value_counts(normalize=True)[nb_bow.classes_].round(4).values,
    "docs en train": y_train.value_counts()[nb_bow.classes_].values,
}, index=nb_bow.classes_)
```

**¿Qué representan P(c) y P(wᵢ|c) en este corpus y cómo los combina el modelo?**

El teorema de Bayes permite invertir una probabilidad condicional: en lugar de preguntar qué tan
probable es un texto dada una categoría, que es lo que se puede estimar contando, permite obtener
qué tan probable es una categoría dado un texto, que es lo que interesa para clasificar.

**P(c)** es la probabilidad a priori de cada categoría, es decir qué tan frecuente es antes de leer
el documento. Se estima como la proporción de noticias de esa categoría dentro del conjunto de
entrenamiento. En este corpus las siete categorías están muy desbalanceadas, de modo que P(c) vale
alrededor de 0.28 para Macroeconomía y apenas 0.023 para Reputación. Esa diferencia funciona como
un sesgo inicial a favor de las categorías mayoritarias, que solo se revierte si el vocabulario del
documento apunta con suficiente fuerza en otra dirección.

**P(w | c)** es la verosimilitud de cada palabra dentro de una categoría: qué tan probable es
encontrar la palabra *wᵢ* si se sabe que la noticia pertenece a la categoría *c*. Se estima
contando cuántas veces aparece esa palabra en todos los documentos de esa categoría en
entrenamiento y dividiendo entre el total de palabras de la categoría. Palabras como *inflación* o
*ipc* tienen una verosimilitud alta bajo Macroeconomía y muy baja bajo Innovación, y son las que
efectivamente discriminan.

Para clasificar un documento nuevo, el modelo evalúa esa expresión para las siete categorías y
devuelve la de mayor valor. El calificativo de *ingenuo* proviene del producto sobre las palabras,
que asume independencia condicional: da por hecho que, conocida la categoría, la aparición de una
palabra no dice nada sobre la aparición de otra. Es un supuesto falso para el lenguaje natural,
donde *banco* y *central* aparecen juntas con mucha más frecuencia de la que predice el producto de
sus probabilidades individuales.

En la implementación se trabaja con logaritmos, de manera que el producto se convierte en una suma
de log-probabilidades. El motivo es numérico: multiplicar cientos de valores menores que uno
produce un resultado que el punto flotante redondea a cero. Se aplica además suavizado de Laplace,
sumando una constante a cada conteo, para que una palabra nunca vista en una categoría reciba una
probabilidad pequeña en lugar de cero, que anularía todo el producto por sí sola.

## 4. Análisis de errores

```python
# Par de categorias que el modelo confunde mas entre si (fuera de la diagonal de la matriz).
clases = nb_bow.classes_
cm_val = confusion_matrix(y_val, nb_bow.predict(Xval_bow), labels=clases)

# Confusion MUTUA: sumamos A->B con B->A y borramos la diagonal (los aciertos)
mutua = cm_val + cm_val.T
np.fill_diagonal(mutua, 0)

i, j = np.unravel_index(np.argmax(mutua), mutua.shape)
cat_a, cat_b = clases[i], clases[j]

print(f"Par mas confundido: {cat_a} <-> {cat_b}  ({mutua[i, j]} errores en total)")
print(f"  {cat_a} clasificadas como {cat_b}: {cm_val[i, j]}")
print(f"  {cat_b} clasificadas como {cat_a}: {cm_val[j, i]}")

pares = [(clases[a], clases[b], int(mutua[a, b]))
         for a in range(len(clases)) for b in range(a + 1, len(clases)) if mutua[a, b] > 0]
pd.DataFrame(sorted(pares, key=lambda p: -p[2]),
             columns=["categoria A", "categoria B", "errores"]).head(6)
```

```python
# Revisar 3 a 5 documentos mal clasificados de ese par y ver que palabras los explican.
pred_val = pd.Series(nb_bow.predict(Xval_bow), index=y_val.index)

mal = ((y_val == cat_a) & (pred_val == cat_b)) | ((y_val == cat_b) & (pred_val == cat_a))
errores = y_val[mal]
print(f"{len(errores)} documentos mal clasificados entre {cat_a} y {cat_b}\n")

terminos_tr = vec_count.get_feature_names_out()
filas_val = list(y_val.index)

def palabras_culpables(idx_doc, n=8):
    """Palabras del documento que mas empujaron hacia la categoria equivocada.
    Peso = conteo en el doc x (log P(w|predicha) - log P(w|real))."""
    real, predicha = y_val[idx_doc], pred_val[idx_doc]
    vector = Xval_bow[filas_val.index(idx_doc)].toarray().ravel()
    fr, fp = list(clases).index(real), list(clases).index(predicha)
    empuje = vector * (nb_bow.feature_log_prob_[fp] - nb_bow.feature_log_prob_[fr])
    top = np.argsort(empuje)[::-1][:n]
    return list(zip(terminos_tr[top], empuje[top].round(2)))

for idx in errores.index[:5]:
    print(f"--- doc {idx} | real: {y_val[idx]}  ->  predicho: {pred_val[idx]}")
    print("   ", df.loc[idx, "texto_norm"][:220], "...")
    print("    palabras que empujaron al error:", palabras_culpables(idx), "\n")
```

```python
# Palabras con mayor P(w|c) para 2 o 3 categorias, desde feature_log_prob_.
def top_verosimilitud(categoria, n=15):
    fila = list(clases).index(categoria)
    idx = np.argsort(nb_bow.feature_log_prob_[fila])[::-1][:n]
    probs = np.exp(nb_bow.feature_log_prob_[fila][idx])
    return [f"{p} ({v:.4f})" for p, v in zip(terminos_tr[idx], probs)]

cats_mostrar = [cat_a, cat_b, "Macroeconomia"]
pd.DataFrame({c: top_verosimilitud(c) for c in cats_mostrar}, index=range(1, 16))
```

Las categorías que más se confunden entre sí son Alianzas y Regulaciones, con 6 errores repartidos

Al revisar los documentos mal clasificados casi siempre hay una palabra suelta haciendo todo el

Ahí se ve el problema del supuesto de independencia condicional. El modelo suma el aporte de cada

entendería que el tema es otro, pero como cada palabra vota sola basta con una para torcer el

Las palabras más probables de cada categoría explican por qué estas dos chocan. Los tops de Alianzas

*mercado*, y solo la primera palabra de cada lista discrimina algo. Macroeconomía es lo contrario,
arranca con *inflación*, *precio* y *tasa*, y no sorprende que sea la que mejor clasifica.


## 5. Efecto de la representación de texto

### 5.1 Implementación propia de Naive Bayes

Los entregables piden comparar `MultinomialNB` contra una implementación propia, así que se
construye el modelo desde cero: P(c) por conteo de documentos, P(wᵢ|c) por conteo de palabras con
suavizado de Laplace, y la decisión en escala logarítmica.

```python
from sklearn.metrics import accuracy_score


class NaiveBayesMultinomial:
    """Naive Bayes multinomial desde cero. alpha = suavizado de Laplace."""

    def __init__(self, alpha=1.0):
        self.alpha = alpha

    def fit(self, X, y):
        X = X.tocsr()
        y = np.asarray(y)
        self.clases_ = np.unique(y)
        n_docs, n_feats = X.shape

        self.log_prior_   = np.zeros(len(self.clases_))
        self.log_verosim_ = np.zeros((len(self.clases_), n_feats))

        for k, c in enumerate(self.clases_):
            Xc = X[y == c]                          # documentos de la clase c

            # log P(c) = log(documentos de la clase / documentos totales)
            self.log_prior_[k] = np.log(Xc.shape[0] / n_docs)

            # log P(w|c): conteo de cada palabra en la clase, + alpha para que nada valga cero
            conteos = np.asarray(Xc.sum(axis=0)).ravel() + self.alpha
            self.log_verosim_[k] = np.log(conteos / conteos.sum())

        return self

    def predict_log_proba(self, X):
        # log P(c) + suma sobre palabras de conteo(w) * log P(w|c).
        # El producto de Bayes se vuelve suma al pasar a logaritmos (evita el underflow).
        return np.asarray(X @ self.log_verosim_.T) + self.log_prior_

    def predict(self, X):
        return self.clases_[self.predict_log_proba(X).argmax(axis=1)]


nb_propio = NaiveBayesMultinomial(alpha=1.0).fit(Xtrain_bow, y_train)

pred_propio  = nb_propio.predict(Xval_bow)
pred_sklearn = nb_bow.predict(Xval_bow)

print("Accuracy implementacion propia:", round(accuracy_score(y_val, pred_propio), 4))
print("Accuracy MultinomialNB        :", round(accuracy_score(y_val, pred_sklearn), 4))
print("Predicciones identicas        :", f"{(pred_propio == pred_sklearn).mean() * 100:.1f}%")

print("\nDiferencia maxima en log P(c)  :",
      np.abs(nb_propio.log_prior_ - nb_bow.class_log_prior_).max())
print("Diferencia maxima en log P(w|c):",
      np.abs(nb_propio.log_verosim_ - nb_bow.feature_log_prob_).max())
```
